***E-Invoice Data for Wholesale marketplace***

This data has the following columns and rows

**Columns** <br>
Invoice_ID - Denotes the invoice Id for the transaction <br>
Seller_Name - Company who sells the services/product <br>
Buyer_Name - Company that buys the services/product <br>
Description - Describes the type of services or product <br>
Quantity - denoted the number of items sold <br>
Unit_price - price per item <br>
Terms_of_payment - Duration in the buyer is expected to pay for the services rendered by the seller <br>
Total_amount - Total amount of the transaction <br>

**Rows**
Unique Transactions which will have all the information denoted above
