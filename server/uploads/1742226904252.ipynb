```
import pandas as pd

# Load the Excel file
file_path = "E-Invoice_Mock_Data.xlsx"  # Update with the correct file path
df = pd.read_excel(file_path, sheet_name="Sheet1")

# Analyze the number of times an item is bought by a specific customer
purchase_frequency = df.groupby(['Buyer_Name', 'Description']).size().reset_index(name='Purchase_Count')

# Display the result
print(purchase_frequency.head())

# Save the analysis as a new Excel file
output_file = "Purchase_Frequency_Analysis.xlsx"
purchase_frequency.to_excel(output_file, index=False)
print(f"Analysis saved to {output_file}")
```
